## 3. Assess data quality

Missing values can affect summaries and visualizations. First, identify where they occur; do not remove rows automatically.

```python
# TODO 1: Count missing values in each column
missing_counts = df.isna().sum()
display(missing_counts)

# TODO 2: Calculate the percentage of missing values in each column
missing_percent = df.isna().mean() * 100
display(missing_percent.round(2))

# TODO 3: Display rows that contain at least one missing value
rows_with_missing = df[df.isna().any(axis=1)]
display(rows_with_missing.head())
```

**Your observations:**  
Missing values occur in some measurement columns and in `sex`; the exact counts are shown above. Missing data can reduce the number of rows available for particular summaries or plots and may affect comparisons if the missingness is not random. Before dropping or imputing values, it would help to know why they are missing and whether the missingness is related to species, island, or other variables.

## 4. Explore numerical variables

Use descriptive statistics to understand the center and spread of a variable. In this section, focus on `body_mass_g`.

```python
# Select body mass and exclude missing values
body_mass = df["body_mass_g"].dropna()

# TODO 1: Calculate the mean and median using NumPy
mean_mass = np.mean(body_mass)
median_mass = np.median(body_mass)
print("Mean:", mean_mass)
print("Median:", median_mass)

# TODO 2: Find the minimum and maximum
print("Minimum:", body_mass.min())
print("Maximum:", body_mass.max())

# TODO 3: Calculate standard deviation and quartiles
print("Standard deviation (population):", np.std(body_mass))
print("Quartiles (25th, 50th, 75th percentiles):", np.percentile(body_mass, [25, 50, 75]))

# TODO 4: Display the five smallest and five largest body-mass values
print("Five smallest body-mass values:")
display(body_mass.nsmallest(5))
print("Five largest body-mass values:")
display(body_mass.nlargest(5))
```

**Your observations:**  
The mean and median are both around 4,000–4,200 g, with the mean somewhat higher than the median. The minimum and maximum show a broad range of body sizes, from roughly 2,700 g to 6,300 g. The extreme values are worth checking against species and other measurements, but they should not be treated as errors without further evidence.
